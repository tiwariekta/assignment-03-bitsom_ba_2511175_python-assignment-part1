```python
#Task 1 — Data Parsing & Profile Cleaning
# Raw Student Data 
raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

for student in raw_students:
    print(student)
```

```python
cleaned_students = []

for student in raw_students:
    # Clean the Name - name has leading/trailing whitespace removed and is converted to Title Case
    cleaned_name = student["name"].strip().title()
    
    # CConvert roll no - roll is converted from a string to an integer.
    roll = int(student["roll"])
    
    # #marks_str is split on ", " and each element is converted to an integer, producing a marks list.
    marks_list = [int(m) for m in student["marks_str"].split(", ")]
    
    # Name Validation Logic:
    is_valid = all(word.isalpha() for word in cleaned_name.split())
    
    # Store cleaned data
    cleaned_students.append({
        "name": cleaned_name,
        "roll": roll,
        "marks": marks_list
    })
    
    # Print a formatted profile card for each cleaned student using f-strings:
    print("=" * 32)
    print(f"Student : {cleaned_name}")
    print(f"Roll No : {roll}")
    print(f"Marks   : {marks_list}")
    
    if is_valid:
        print("✓ Valid name")
    else:
        print("✗ Invalid name")
    
    print("=" * 32)
```

```python
# print the name in ALL CAPS and lowercase for the student with roll number 103.
for student in cleaned_students:
    if student["roll"] == 103:
        name = student["name"]
        print(name.upper())
        print(name.lower())
```

```python
#Task 2 — Marks Analysis Using Loops & Conditionals (8 marks)
# Data
student_name = "Ayesha Sharma"
subjects     = ["Math", "Physics", "CS", "English", "Chemistry"]
marks        = [88, 72, 95, 60, 78]
```

```python
# 1: Using a for loop, print each subject alongside its marks and a grade label based on this scheme:
# Grading Logic
for subject, mark in zip(subjects, marks):
    
    if mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    else:
        grade = "F"
    
    print(f"{subject}: {mark} ({grade})")

# 2: Total, average, highest & lowest mark calculations
total = sum(marks)
average = round(total / len(marks), 2)

print("\nTotal:", total)
print("Average:", average)

max_mark = max(marks)
min_mark = min(marks)

print("Highest:", subjects[marks.index(max_mark)], max_mark)
print("Lowest:", subjects[marks.index(min_mark)], min_mark)

# 3: Using a while loop, simulate a marks-entry system that allows adding new subjects
new_entries = []
new_count = 0

while True:
    subject = input("\nEnter subject name (or 'done' to stop): ")
    
    # End loop if input is done
    if subject.lower() == "done":
        break
    
    # Inner loop to iterate until valid mark is input and to preserve the subject and not ask repeatedly
    while True:
        marks_input = input("Enter marks (0-100): ")
        
        # Input value validation for marks to be numberic 
        if not marks_input.isdigit():
            print("Warning! Invalid input. Enter numeric value.")
            continue
        
        marks_value = int(marks_input)
        
        # Input value validation for marks to be in the range 0-100 
        if marks_value < 0 or marks_value > 100:
            print("Warning! Marks should be between 0 and 100.")
            continue
        break      

    # Enter new subject only after valid marks enetered
    subjects.append(subject)
    marks.append(marks_value)

    # store new entry
    new_entries.append((subject, marks_value))
    new_count += 1

# New Subjects Added
print("\nNew subjects added:", new_count)
if new_entries:
    print("\nNewly Added Subjects:")
    for sub, mark in new_entries:
        print(f"{sub}: {mark}")
else:
    print("No new subjects were added.")

# recomputing old+new subjects avg 
new_average = round(sum(marks) / len(marks), 2)
print("Updated Average:", new_average)
```

```python
# Task 3 — Class Performance Summary
class_data = [
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]
```

```python
# Formatted Class Report Header
print("Name               | Average | Status")
print("-" * 40)

pass_count = 0
fail_count = 0
topper_name = ""
topper_avg = 0
total_avg_sum = 0

# Loop through class_data and for each student & computing average, result status
for name, marks in class_data:
    
    avg = round(sum(marks) / len(marks), 2)
    
    # Pass/Fail - to identify number of students who passed and who failed
    if avg >= 60:
        status = "Pass"
        pass_count += 1
    else:
        status = "Fail"
        fail_count += 1
    
    # Print report row
    print(f"{name:<18} | {avg:>7} | {status}")
    
    # Track Class topper
    if avg > topper_avg:
        topper_avg = avg
        topper_name = name
    
    # Track class avg
    total_avg_sum += avg

# Summary
print("\nSummary:")
print("Passed:", pass_count)
print("Failed:", fail_count)
print("Topper:", topper_name, topper_avg)

class_avg = round(total_avg_sum / len(class_data), 2)
print("Class Average:", class_avg)
```

```python
# Task 4 — String Manipulation Utility
# Initial Data
essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "
```

```python
# Strip leading and trailing whitespace from essay. Store this as clean_essay and use it for all steps below.
clean_essay = essay.strip()

print("1. Clean Essay:")
print(clean_essay)

# Convert clean_essay to Title Case and print it.
title_case = clean_essay.title()

print("\n2. Title Case:")
print(title_case)

# Count how many times the word "python" appears in clean_essay (case-insensitive). Print the count.
count_python = clean_essay.count("python")

print("\n3. Count of 'python':", count_python)

# Replace every occurrence of "python" in clean_essay with "Python 🐍" and print the result.
replaced_text = clean_essay.replace("python", "Python 🐍")

print("\n4. Replaced Text:")
print(replaced_text)

# Split clean_essay into individual sentences by splitting on ". " (a period followed by a space). Print the resulting list.
sentences = clean_essay.split(". ")

print("\n5. Sentences List:")
print(sentences)

# Print each sentence on a new line, numbered starting from 1. Add a "." at the end of each sentence if it does not already end with one.
print("\n6. Numbered Sentences:")

for i, sentence in enumerate(sentences, start=1):
    
    # Convert to Title Case
    sentence = sentence.title()
    
    # Ensure sentence ends with "."
    if not sentence.endswith("."):
        sentence += "."
    
    print(f"{i}. {sentence}")
```
